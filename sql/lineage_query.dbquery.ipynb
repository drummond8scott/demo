```sql
select t.table_schema, t.table_name
from learn_tuva.information_schema.tables t
left join system.access.table_lineage l
  on l.source_table_full_name = 'learn_tuva.' || t.table_schema || '.' || t.table_name
where l.source_table_full_name is null
  and t.table_schema <> 'information_schema'
order by t.table_schema, t.table_name;
```
